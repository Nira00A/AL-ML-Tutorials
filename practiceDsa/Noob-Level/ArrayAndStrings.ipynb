```python
# Double Linked List

class DoubleLinked:
    def __init__(self , val , next=None , prev=None):
        self.next = next
        self.prev = prev
        self.val = val

    def __str__(self):
        return (str(self.val))

def traverse(Head):
    curr = Head
    elements = []
    while curr:
        elements.append(str(curr))
        curr = curr.next
    return " <-> ".join(elements)

Head = DoubleLinked(2)

def insertStart(Head , val):
    new_val = DoubleLinked(val)
    if Head:
        new_val.next = Head
        Head.prev = new_val

    return new_val
    
head = DoubleLinked(2)

head = insertStart(head , 12)
head = insertStart(head , 13)
head = insertStart(head , 14)

print(traverse(head))
```

```python
#Question2 - The Problem: You have an array of prices. Find the max profit 
# (buy low, sell high).You must buy before you sell.
prices = [7,6,4,3,1]


def maxProfit(prices: List[int]):
    min_price = float('inf')
    diff = 0

    for n in prices:
        if min_price > n:
            min_price = n

        elif (n - min_price) > diff:
            diff = n - min_price

    return diff

print(maxProfit(prices))
```

```python
#Question3 - The Problem: Given a string, determine if it is a palindrome 
# (reads the same forward and backward), considering only alphanumeric 
# characters and ignoring cases. (LeetCode #125)

def palindrome(string: str):
    length = len(string)
    l = 0
    r = length - 1
    
    while l < r:
        if string[l] != string[r]:
            return False
        
        l+= 1
        r-= 1

    return True

string = 'aabAa'
print(palindrome(string))

```

```python
#Question4 - The Problem: Find the contiguous subarray 
# (containing at least one number)
#  which has the largest sum.

def kadaneAlgo(arr: List[int]):
    max_so_far = arr[0]
    current_sum = arr[0]

    for i in range(1 , len(arr)):
        current_sum = max(arr[i] , current_sum + arr[i])

        max_so_far = max(max_so_far, current_sum)
        
    return max_so_far
```

```python
#Recurtion practice Merge Sort

def mergeSort(arr: List[int]):
    if len(arr) <= 1:
        return arr
    
    # Divide
    mid = len(arr) / 2
    leftHalf = arr[:mid]
    rightHalf = arr[mid:]

    # Conquer
    sortedLeft = mergeSort(leftHalf)
    sortedRight = mergeSort(rightHalf)

    # Combine
    return merge(sortedLeft , sortedRight)

def merge(left: List[int] , right: List[int]):
    i:int = 0
    j: int = 0
    result: List[int] = []

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:0])
    result.extend(right[j:0])
    return result
```

```python
#Problem4 - The Goal: Calculate $x^n$ (e.g., $2^{10}$).
# The Constraint: Do it faster than $O(N)$. You want $O(\log N)$. (LeetCode #50)

def exN(x:int , n:int):
    if(n == 0):
        return 1
    
    half = exN(x,n //2)
    
    if n % 2 == 0:
        return half * half
    else:
        return x * half * half
    

print(exN(2 , 2))
```

## Problem 1: Product of Array Except Self
### The Challenge: Given an integer array nums, return an array answer such that answer[i] is equal to the product of all the elements of nums except nums[i].

Crucial Constraint: You cannot use the division operator /, and you must solve it in $O(N)$ time.Example: nums = [1, 2, 3, 4]Output: [24, 12, 8, 6]  (e.g., at index 2: $1 \times 2 \times 4 = 8$)Interview Hint: If you can't divide the total product by the current number, calculate the prefix products (everything to the left of i) and suffix products (everything to the right of i). Can you do this using just a single output array to achieve $O(1)$ auxiliary space?

```python
from typing import List

nums = [1,2,3,4]

def arrayExcept(arr: List[int]) -> List[int] :
    n = len(arr)
    ans = [1] * n

    for i in range(1 , n):
        ans[i] = ans[i - 1] * arr[i - 1]

    right_prod = 1
    for i in range(n - 1, -1 , -1):
        ans[i] *= right_prod
        right_prod *= nums[i]

    return ans

print(arrayExcept(nums))
```

## Problem 2: Container With Most WaterThe 
### Challenge: You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the $i^{th}$ line are (i, 0) and (i, height[i]). Find two lines that together with the x-axis form a container, such that the container contains the most water. Return the maximum amount of water a container can store.  

Example: height = [1, 8, 6, 2, 5, 4, 8, 3, 7]

Output: 49 (Area formed by lines at index 1 and index 8: $\text{width} = 8 - 1 = 7$; $\text{height} = \min(8, 7) = 7$; $\text{Area} = 7 \times 7 = 49$)

Interview Hint: Brute force takes $O(N^2)$. To get to $O(N)$, place two pointers at the absolute extremes of the array (left and right). To maximize area, you need width and height. If you move a pointer, which one should you move inward—the taller line or the shorter line?

```python
def max_area(arr: List[int]) -> int:
    lp = 0
    rp = len(arr) - 1
    max_water = 0

    while lp < rp:
        width = rp - lp
        height = min(arr[lp] , arr[rp])

        curr_area = width * height

        max_water = max(max_water , curr_area)

        if arr[lp] < arr[rp]:
            lp += 1
        else:
            rp -= 1

    return max_water

print(max_area([1,8,6,2,5,4,8,3,7]))
```

```python
def string_compress(arr: List):
    read = 0
    write = 0
    n = len(arr)

    while read < n:
        char = arr[read]
        count = 0

        while read < n and arr[read] == char:
            read += 1
            count += 1

        arr[write] = char
        write += 1

        if count > 1:
            for digit in str(count):
                arr[write] = digit
                write += 1

    return write

print(string_compress(["a", "a", "b", "b", "b", "c", "c", "c"]))
```

```python
def compress(chars: list[str]) -> int:
    read = 0
    write = 0
    n = len(chars)
    
    while read < n:
        char = chars[read]
        count = 0
        
        # 1. Count how many times this character repeats consecutively
        while read < n and chars[read] == char:
            read += 1
            count += 1
            
        # 2. Write the character to our compressed slot
        chars[write] = char
        write += 1
        
        # 3. If it repeated, write the count (handling multi-digit numbers)
        if count > 1:
            for digit in str(count):
                chars[write] = digit
                write += 1
                
    # write now represents the new length of the compressed array
    return write
```
